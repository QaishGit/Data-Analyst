### 5 Number Summary and Box Plot

```python
import numpy as np

lst_mrks = [89, 56, 78, 86, 98, 56, 74, 53, 84, 93, 71, 65, 91, 92, 72, 63, 80]
minimum, Q1, median, Q3, maximum = np.quantile(lst_mrks, [0, 0.25, 0.50, 0.75, 1.0])

print('minimum: ', minimum)
print('Q1: ', Q1)
print('median: ', median)
print('Q3: ', Q3)
print('maximum: ', maximum)
```

```python
IQR = Q3 - Q1
print('IQR: ', IQR)
```

```python
lower_fence = Q1 - (1.5 * IQR)
higher_fence = Q3 + (1.5 * IQR)

print('lower_fence: ', lower_fence)
print('higher_fence: ', higher_fence)
```

```python
## Creating Box Plot(without outliers)
import seaborn as sns

sns.boxplot(lst_mrks)
```

```python
## Creating Box Plot(with outliers)

lst_mrks = [89, 56, 78, 86, 98, 56, 74, 53, 84, 93, 71, 65, 91, 92, 72, 63, 80, 120, 125, 140, 145]

sns.boxplot(lst_mrks)
```
