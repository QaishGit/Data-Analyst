```python
## 1
matrix = np.random.randint(1, 11, size=(3, 3))
print("Matrix: ")
print(matrix)

## Determinant
determinant = np.linalg.det(matrix)
print(f"\nDeterminant: {determinant}")

## Inverse
inverse = np.linalg.inv(matrix)
print("\nInverse of matrix:")
print(inverse)

## Eigenvalues
eigenvalues = np.linalg.eigvals(matrix)
print(f"\nEigenvalues: {eigenvalues}")

## Eigenvalues and Eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(matrix)
print(eigenvalues)
print(eigenvectors)
```

```python
## 1
data_type = [('name', 'U10'), ('age', 'i4'), ('weight', 'f4')]
data = np.array([("Alice", 20, 56.8), ("Bob", 25, 70.5), ("Charlie", 30, 78.45)], dtype=data_type)
print("Original array:")
print(data)
print(type(data))

sorted_data = np.sort(data, order='age')
print("\nSorted array on the basis of age:")
print(sorted_data)

sorted_weight = np.sort(data, order='weight')
print("\nSorted array on the basis of weight")
print(sorted_weight)
```

```python
## 2
data_type = [('x', 'i4'), ('y', 'i4')]
coordinates = np.array([(1, 2), (3, 4), (5, 6)], dtype=data_type)
print("Cordinates: ", coordinates)

euclid_distance = np.sqrt((coordinates['x'][:, np.newaxis] - coordinates['x'])**2 + \
                          (coordinates['y'][:, np.newaxis] - coordinates['y'])**2)
print("\nEuclidean distance:")
print(euclid_distance)
                          
```

```python
## 1
import numpy.ma as ma
array = np.random.randint(1, 16, size=(4, 4))
print("Original array:")
print(array)

masked_array = ma.masked_greater(array, 10)
print("\nMasked array:")
print(masked_array)

unmasked_sum = masked_array.sum()
print("\nSum of Unmasked elements:", unmasked_sum)
```

```python
## 2
array = np.random.randint(1, 21, size=(3, 3))
print("Original array")
print(array)

masked_array = ma.masked_array(array, mask=np.eye(3, dtype=bool))
print("\nMasked array:")
print(masked_array)

unmasked_mean = masked_array.mean()
masked_array = masked_array.filled(unmasked_mean)

print("\nModified array:")
print(masked_array)
```
