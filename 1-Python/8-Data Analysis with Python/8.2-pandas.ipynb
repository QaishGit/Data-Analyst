### Pandas - DataFrames and Series

Series

```python
## Creating series using a list

import pandas as pd

data = [1, 2, 3, 4, 5]

series = pd.Series(data)
print("Series:\n", series)
```

```python
## Creating Series using a dictionary

data = {'a': 1, 'b': 2, 'c': 3}

series_dict = pd.Series(data)
print(series_dict)
```

```python
## Creating ou own index values

data = [10, 20 ,30]
index = ['a', 'b', 'c']
pd.Series(data, index = index)
```

DataFrame

```python
## Creatind DataFrame from a dictionary of list

data = {
    "Name": ["Krish", "John", "Jack"],
    "Age": [25, 30 ,35],
    "City": ["Bangalore", "New York", "Florida"]
}

df = pd.DataFrame(data)
print(df)
print(type(df))
```

```python
## Converting above DataFrame into an array using numpy

import numpy as np

np.array(df)
```

```python
## Creating a DataFrame from list of dictionaries

data = [
    {"Name": "Krish", "Age": 30, "City": "Bangalore"},
    {"Name": "John", "Age": 35, "City": "New York"},
    {"Name": "Jack", "Age": 40, "City": "Florida"}
]
df = pd.DataFrame(data)
print(df)
```
