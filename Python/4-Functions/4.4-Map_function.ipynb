### map function

```python
def square(num):
    return num**2

square(4)
```

```python
## using map function

numbers = [1, 2, 3, 4, 5, 6, 7]

squares = list(map(square, numbers))
print(squares)
```

```python
## Using map and lambda functions

squares_1 = list(map(lambda x: x**2, numbers))
print(squares_1)
```

Mapping multiple iterables

```python
num1 = [1, 2, 3]
num2 = [4, 5, 6]

add_numbers = list(map(lambda x, y: x + y, num1, num2))
print(add_numbers)
```

```python
## Using map() to convert string of numbers into int

str_num = ['1', '2', '3', '4', '5']
int_num = list(map(int, str_num))
print(int_num)
```

```python
words = ['apple', 'banana', 'mango']
upper_word = list(map(str.upper, words))

print(upper_word)
```

```python
## using dictionary

def get_name(person):
    return person['name']

people = [
    {"name": "krish", "age": 32},
    {"name": "cash", "age": 25}
]

list(map(get_name, people))
```

```python
## Above code using lamda

list(map(lambda x: x['name'], people))
```
