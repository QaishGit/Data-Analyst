### Lambda Function


```python
# Using normal function
def add(a, b):
    return a + b

add(2, 3)
```

```python
# Using lambda function
add_1 = lambda a, b: a+b
add_1(2, 3)
```

```python
# Normal function
def even(num):
    if num%2==0:
        return True
    else:
        return False
even(3)
```

```python
# Lambda function
even_1 = lambda x: x%2==0
even_1(2)
```

map()

```python
numbers = [1, 2, 3, 4, 5]

map(lambda x: x**2, numbers)
```

```python
list(map(lambda x: x**2, numbers))
```
