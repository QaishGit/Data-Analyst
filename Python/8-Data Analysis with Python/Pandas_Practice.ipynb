Question 1:

```python
## Answer
import pandas as pd

df = pd.DataFrame({
    'Name': ['Jack', 'Rahul', 'Sara', 'John', 'Rohan'],
    'Age': [21, 22, 20, 23, 21],
    'Marks': [85, 72, 91, 68, 88],
   'City': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai']
})
print("DataFrame:")
print(df)

print("\nFirst three rows:")
print(df.head(3))

print()
print(df[['Name', 'Marks']]) ## Double square brackets because we are selecting multiple columns

print(f"\nAnerage marks: {df['Marks'].mean()}")

print("\nStudent with highest marks:")
print(df.loc[df['Marks'].idxmax()])

print()
print(df[df['Marks']>80])
```

Question 2:

```python
df = pd.DataFrame({
    'Employee': ['Amit', 'Neha', 'Raj', 'Sara', 'John'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT'],
    'Salary': [55000, 45000, 70000, 60000, 80000],
   'Experince': [2, 3, 5, 4, 6]
})
print("DataFrame:")
print(df)

## DataFrames's shape
print("\nShape of DataFrame: ", df.shape)

## Average Salary
print("\nAverage Salary: ", df['Salary'].mean())

## Employee with lowest salary
print("\nEmployee with lowest salary: ", df.loc[df['Salary'].idxmin()]['Employee'])

## Employees from IT Department
print("\nEmployees from IT Department:")
print(df[df['Department']=='IT'])

## Sorting Salary descending
print("\nSorted DataFrame on Salary in descending order:")
df.sort_values('Salary', ascending=False)
```
