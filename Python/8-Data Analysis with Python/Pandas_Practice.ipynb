### Basics

Question 1:

```python
## Answer

df = pd.DataFrame({
    'Name': ['Jack', 'Rahul', 'Sara', 'John', 'Rohan'],
    'Age': [21, 22, 20, 23, 21],
    'Marks': [85, 72, 91, 68, 88],
   'City': ['Delhi', 'Mumbai', 'Delhi', 'Pune', 'Mumbai']
})
print("DataFrame:")
print(df)

print("\nFirst three rows:")
print(df.head(3))

print()
print(df[['Name', 'Marks']]) ## Double square brackets because we are selecting multiple columns

print(f"\nAnerage marks: {df['Marks'].mean()}")

print("\nStudent with highest marks:")
print(df.loc[df['Marks'].idxmax()])

print()
print(df[df['Marks']>80])
```

Question 2:

```python
## Answer

df = pd.DataFrame({
    'Employee': ['Amit', 'Neha', 'Raj', 'Sara', 'John'],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT'],
    'Salary': [55000, 45000, 70000, 60000, 80000],
   'Experince': [2, 3, 5, 4, 6]
})
print("DataFrame:")
print(df)

## DataFrames's shape
print("\nShape of DataFrame: ", df.shape)

## Average Salary
print("\nAverage Salary: ", df['Salary'].mean())

## Employee with lowest salary
print("\nEmployee with lowest salary: ", df.loc[df['Salary'].idxmin()]['Employee'])

## Employees from IT Department
print("\nEmployees from IT Department:")
print(df[df['Department']=='IT'])

## Sorting Salary descending
print("\nSorted DataFrame on Salary in descending order:")
df.sort_values('Salary', ascending=False)
```

Question 3:

```python
## Answer

df = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Chair', 'Table', 'Monitor', 'Desk'],
    'Category': ['Electronics', 'Electronics', 'Furniture', 'Furniture', 'Electronics', 'Furniture'],
    'Sales': [60000, 90000, 20000, 35000, 45000, 25000],
    'Quantity': [3, 6, 5, 3, 5, 4]
})
print("DataFrame:")
print(df)

## Total Sales
print(f"\nTotal Sales: {df['Sales'].sum()}")

## Average Sales
print("\nAverage Sales: ", df['Sales'].mean())

## Total Sales for each category
category_wise_sales = df.groupby('Category')['Sales'].sum()
print("\nTotal sales for each category:")
print(category_wise_sales)

## Product with highest sales
print("\nProduct with highest sales:", df.loc[df['Sales'].idxmax()]['Product'])

## Add new column 'Sales_per_Item
df['Sales_per_Item'] = df['Sales']/df['Quantity']
print('\nModified DataFrame:')
print(df)
```

Question 4:

```python
## Answer 

df = pd.DataFrame({
    'Name': ['Jack', 'Sara', 'John', 'David', 'Hugh'],
    'Age': [24, None, 27, 25, None],
    'Salary': [50000, 60000, None, 55000, 70000],
    'Department': ['IT', 'HR', 'IT', None, 'IT']
})
print("Original DataFrame:")
print(df)

## Number of missing values in each column
print("\nNumber of missing values in each column:")
print(df.isnull().sum())

## Total number of missing values
print("\nTotal number of missing values:")
print(df.isnull().sum().sum())

## Replacing missing Age values with average age
print("\nModified DataFrame with replaced missing Age values with average age values:")
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)

## Replacing missing Salary values with median of salary
print("\nModified DataFrame with replaced missing Salary values with median of Salary:")
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
print(df)

## Replacing missing Department values with "Unknown"
print("\nModified DataFrame with replaced missing Department values with 'Unknown':")
df['Department'] = df['Department'].fillna('Unkown')
print(df)
```

Question 5:

```python
## Answer

df = pd.DataFrame({
    'Employee': ['A', 'B', 'C', 'D', 'E', 'F', 'G'],
    'Department': ['IT', 'IT', 'HR', 'HR', 'Finance', 'Finance', 'IT'],
    'Salary': [60000, 75000, 45000, 50000, 70000, 80000, 90000],
    'Experience': [2, 5, 3, 4, 6, 8, 7]
})
print("Original DataFrame:")
print(df)

## Average Salary of each Department
print("\nAverage salary of each Department:")
avg_salary = df.groupby('Department')['Salary'].mean()
print(avg_salary)

## Maximum Salary of each Department
print("\nMaximum salary of each Department:")
max_salary = df.groupby('Department')['Salary'].max()
print(max_salary)

## Minimum Salary of each Department
print("\nMinimum salary of each Department:")
min_salary = df.groupby('Department')['Salary'].min()
print(min_salary) 

## Average Experience of each Department
print("\nAverage experience of each Department:")
avg_exp = df.groupby('Department')['Experience'].mean()
print(avg_exp)

## The Department with highest average salary
print("\nDepartment with highest average salary:")
print(df.groupby('Department')['Salary'].mean().idxmax())
print()
## OR
print(avg_salary[avg_salary==avg_salary.max()])
```
