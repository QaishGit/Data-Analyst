```python
## 1
import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.randint(1, 101, size=(6, 4)), columns=['A', 'B', 'C', 'D'])
print("Original DataFrame:")
print(df)

df.set_index('A', inplace=True)
print("\nModified DataFrame with changed index:")
print(df)
```

```python
## 2

df = pd.DataFrame(np.random.randint(1, 10, size=(3, 3)), index=['X', 'Y', 'Z'], columns=['A', 'B', 'c'])
print("Original DataFrame:")
print(df)

print("\nElement at row 'Y' and column 'B': ", df.at['Y', 'B']) 
print("\nElement at row 'Y' and column 'B': ", df.loc['Y']['B']) # OR df.loc['Y', 'B']
```

```python
## 1
df = pd.DataFrame(np.random.randint(1, 100, size=(4, 3)), columns=['A', 'B', 'C'])
print("Original DataFrame:")
print(df)

df.iat[0, 1] = np.nan
df.iat[1, 2] = np.nan
df.iloc[2, 0] = np.nan
print("\nDataFarem with NaN values")
print(df)

df.fillna(df.mean(), inplace=True)
print("\nModified DataFrame with filled missing values with mean:")
print(df)
```

```python
## 2

df = pd.DataFrame(np.random.randint(1, 100, size=(6, 4)), columns=['A', 'B', 'C', 'D'])
print("Original DataFrame:")
print(df)

df.iat[0, 3] = np.nan
df.iloc[3, 2] = np.nan
df.iat[5, 1] = np.nan
df.iloc[1, 0] = np.nan
print("\nDataFrame with missing values:")
print(df)

df.dropna(inplace=True)
print("\nModified DataFrame:")
print(df)
```

```python
## 1

df = pd.DataFrame({'Category': np.random.choice(['A', 'B', 'C'], size=10),
                   'Value': np.random.randint(1, 100, size=10)})
print("Original DataFrame:")
print(df)

grouped = df.groupby('Category')['Value'].agg(['sum', 'mean'])
print("\nGrouped DataFrame:")
print(grouped)
```

```python
## 2

df = pd.DataFrame({'Product': np.random.choice(['Product1', 'Product2', 'Product3'], size=10),
                   'Category': np.random.choice(['A', 'B', 'C'], size=10),
                   'Sales': np.random.randint(1, 100, size=10)})
print("Original DataFrame:")
print(df)

grouped_category = df.groupby('Category')['Sales'].sum()
print("\nGrouped DataFrame:")
print(grouped_category)
```

Assignment 5

```python
## 1

df1 = pd.DataFrame({'Key': ['A', 'B', 'C', 'D'], 'Value1': np.random.randint(1, 100, size=4)})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D', 'E'], 'Value2': np.random.randint(1, 100, size=4)})
print("DataFrame 1:")
print(df1)
print("\nDataFrame 2:")
print(df2)

merge_df = pd.merge(df1, df2, on='Key', how='inner')
print("\nInner join of 2 DataFrames:")
print(merge_df)
```

```python
## 2

df1 = pd.DataFrame({'A': np.random.randint(1, 100, size=3), 'B': np.random.randint(1, 100, size=3)})
df2 = pd.DataFrame({'C': np.random.randint(1, 100, size=3), 'D': np.random.randint(1, 100, size=3)})
print("First DataFrame:")
print(df1)
print("\nSeconda DataFrame:")
print(df2)

concat_rows = pd.concat([df1, df2], axis=0)
print("\nRow-wise cocatination:")
print(concat_rows)

concat_columns = pd.concat([df1, df2], axis=1)
print("\nColumn-wise Concatination:")
print(concat_columns)
```
