File Handling - reading and writing in file

```python
## Reading a file

with open('example.txt', 'r') as file:
    content = file.read()
    print(content)
```

```python
## Reading a file line by line

with open('example.txt', 'r') as file:
    for line in file:
        print(line.strip())
```

```python
## Writing in a file(Overwriting)

with open('example.txt', 'w') as file:
    file.write("Hello World!\n")
    file.write("This is a new line")
```

```python
## Writing a file(without overwriting)

with open('example.txt', 'a') as file:
    file.write("\nAppending a new line")
```

```python
## Writing a list of lines to a file

lines = ["\nFirst line\n", "Second line\n", "Third line\n"]
with open('example.txt', 'a') as file:
    file.writelines(lines)
```

Binary Files

```python
## Writing in a Binary file

data = b'\x00\x01\x02\x03\x04'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
## Reading a Binary file

with open('example.bin', 'rb') as file:
    content = file.read()
    print(content)
```

```python
## Reading from a source file and then writing it to a destination file

with open('example.txt', 'r') as source_file:
    content = source_file.read()

with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
## Read a text file and return the number of lines, words and characters in it

def count_text_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        count_lines = len(lines)
        count_words = sum(len(line.split()) for line in lines)
        count_chars = sum(len(line.replace(" ", "")) for line in lines)

    return count_lines, count_words, count_chars

lines, words, characters = count_text_file('example.txt')
print(f"The file contains {lines} lines, {words} words, and {characters} characters")
```

```python
## Writing and then reading

with open('example.txt', 'w+') as file:

    ## writing in the file
    file.write("Hello World\n")
    file.write("This is new line\n")

    ## Moving cursor back from line 2 to line 1
    file.seek(0)

    ## now reading the file 
    content = file.read()
    print(content)
```
